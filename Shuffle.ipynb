Create a program that simulates picking a random card from a deck.  

```python
import random

suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
ranks = [
    "2", "3", "4", "5", "6", "7", "8", "9", "10",
    "Jack", "Queen", "King", "Ace"
]


deck = [(rank, suit) for suit in suits for rank in ranks]


card = random.choice(deck)

# Display the result
print("Randomly picked card:", card[0], "of", card[1])
```

Given a list of numbers, shuffle the even and odd numbers separately and then merge them.  

```python
import random
lis=[1,2,3,4,5,6,88,99]
random.shuffle(lis)
a=[x for x in lis if x%2==0]
b=[x for x in lis if x%2!=0]
a.extend(b)
print(a)
```

Given two lists, shuffle them together such that one element from the first list is followed by one from the second, and so on.  

```python
import random

list1 = [1, 2, 3, 4]
list2 = ['a', 'b', 'c', 'd']

random.shuffle(list1)
random.shuffle(list2)

result = []

for x, y in zip(list1, list2):
    result.append(x)
    result.append(y)

print(result)
```

Given a list of numbers, shuffle them such that no two even numbers or two odd numbers come together.  

```python
import random

numbers = [1, 2, 3, 4, 5, 6]

even = [x for x in numbers if x % 2 == 0]
odd = [x for x in numbers if x % 2 != 0]

random.shuffle(even)
random.shuffle(odd)

result = []

for e, o in zip(even, odd):
    result.append(e)
    result.append(o)

print(result)
```

Given a list of names, create random pairs

```python
import random

names = ["Ram", "Sita", "John", "David", "Priya", "Rahul"]

random.shuffle(names)

pairs = []

for i in range(0, len(names), 2):
    pair = (names[i], names[i + 1])
    pairs.append(pair)

print(pairs)
```

Given a string, shuffle its characters but ensure vowels remain in the same position.

```python
import random

text = "hello"

# Find consonants
consonants = [char for char in text if char.lower() not in "aeiou"]

# Shuffle consonants
random.shuffle(consonants)

# Replace only consonant positions
result = []
index = 0

for char in text:
    if char.lower() in "aeiou":
        result.append(char)
    else:
        result.append(consonants[index])
        index += 1

print("".join(result))
```

Given a list of numbers, shuffle it such that no number is next to its original neighbors.

```python
import random

numbers = [1, 2, 3, 4, 5, 6]

original = numbers.copy()

# Store original neighboring pairs
neighbors = set()

for i in range(len(original) - 1):
    neighbors.add((original[i], original[i + 1]))
    neighbors.add((original[i + 1], original[i]))

while True:
    random.shuffle(numbers)

    valid = True

    for i in range(len(numbers) - 1):
        if (numbers[i], numbers[i + 1]) in neighbors:
            valid = False
            break

    if valid:
        break

print("Original:", original)
print("Shuffled:", numbers)
```
