Write a Python program to find and print the first prime number greater than 100. Use a loop and the break statement to exit the loop once you find the prime number.  


```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

for number in range(101, 1000):
    if is_prime(number):
        print(f"The first prime number greater than 100 is: {number}")
        break
```

Create a Python program to calculate the sum of all numbers from 1 to 100. However, if the number is divisible by both 3 and 5, skip it using the continue statement.  

```python
sum=0
for i in range(1,101):
  if i%3==0 and i%5==0:
    continue
  else:
    sum+=i
print(sum)
```

Implement a program that calculates the average of a list of numbers, but ignore any negative numbers. Use the continue statement to skip negative numbers in the list.

```python
lis=[1,4,8,9,-1,-5,-7,5,5,7]
sum=0
for i in range(len(lis)+1):
  if i<0:
    continue
  else:
    sum+=i

avg=sum/len(lis)
print(avg)
```

Write a Python program that finds the first 10 even numbers. If a number is divisible by 3, skip it using the continue statement.

```python
lst=[]
i=1
while len(lst)<=10:
  if i%2==0 and i%3!=0:
    lst.append(i)
  i+=1
print(lst)
```

Write a Python program that generates and prints the first 20 terms of the Fibonacci sequence. However, if a term is greater than 100, stop generating the sequence using the break statement.  

```python
lst=[0,1]
for i in range(2,20):
  c=lst[-1]+lst[-2]
  if c>100:
    break
  else:
    lst.append(c)

print(lst)
```
