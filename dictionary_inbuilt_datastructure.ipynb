Checking for the existence of a key and getting its value with a default:  

```python
my_dict = {'name': 'Alice', 'age': 30}

# Get value for an existing key
name = my_dict.get('name', 'N/A')
print(f"Name: {name}")

# Get value for a non-existing key with a default
occupation = my_dict.get('occupation', 'Unemployed')
print(f"Occupation: {occupation}")

# Get value for a non-existing key without specifying a default (returns None)
salary = my_dict.get('salary')
print(f"Salary: {salary}")
```

Merging dictionaries so that values are added together if keys exist in both dictionaries:  

```python
dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = {'b': 5, 'c': 4, 'd': 6}

merged_dict = dict1.copy() # Start with a copy of dict1

for key, value in dict2.items():
    if key in merged_dict: # If key exists in both, add values
        merged_dict[key] += value
    else: # Otherwise, add the new key-value pair
        merged_dict[key] = value

display(merged_dict)
```

 Implementing a switch-case statement using a dictionary:  

```python
def play_game():
    return "Starting the game..."

def show_settings():
    return "Opening settings menu..."

def exit_app():
    return "Goodbye!"

def default_action():
    return "Invalid choice. Please try again."

# 1. Define the switch-case dictionary
switch = {
    '1': play_game,
    '2': show_settings,
    '3': exit_app
}

# 2. Simulate user input
user_choice = input('enter the option 1 , 2, 3 : ')

# 3. Get the function and execute it immediately
# The second argument acts as the 'default' case
action = switch.get(user_choice, default_action)
result = action()

print(result)
# Output: Opening settings menu...
```

Using a dictionary to implement a simple calculator:  

```python
def add(x,y):
  return x+y
def sub(x,y):
  return x-y
def div(x,y):
  if y==0:
    print('cannot be divided by zero')
  else:
    return x//y
def multi(x,y):
  return x*y

switch={
    '1':add,
    '2':sub,
    '3':div,
    '4':multi
}

x=int(input('enter the number: '))
y=int(input('enter the number: '))

n=input('choose the options 1,2,3,4 :')

action=switch.get(n,lambda x,y:'choose valid option')
result=action(x,y)
print(result)
```

How can you create a deep copy of a nested dictionary?

```python
import copy

original_nested_dict = {
    'a': 1,
    'b': {'inner_key': 10, 'another_inner': [1, 2, 3]},
    'c': [4, 5]
}

deep_copy_dict = copy.deepcopy(original_nested_dict)

# Modify the deep copy
deep_copy_dict['a'] = 100
deep_copy_dict['b']['inner_key'] = 200
deep_copy_dict['c'].append(6)

print(f"Original dictionary: {original_nested_dict}")
print(f"Deep copy dictionary: {deep_copy_dict}")

```

How can you create a dictionary that tracks the execution time of functions?

```python
import time

# Dictionary to store function execution times
function_execution_times = {}

def track_time(func):
    """A decorator that tracks the execution time of the decorated function."""
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        execution_time = end_time - start_time
        function_execution_times[func.__name__] = execution_time
        print(f"Function '{func.__name__}' executed in {execution_time:.4f} seconds")
        return result
    return wrapper

# Example functions to demonstrate the decorator

@track_time
def long_running_task(iterations):
    """Simulates a task that takes some time."""
    total = 0
    for _ in range(iterations):
        total += 1
    return total

@track_time
def another_task(delay_seconds):
    """Simulates another task with a delay."""
    time.sleep(delay_seconds)
    return "Task completed!"

# Call the decorated functions
long_running_task(10000000)
another_task(0.5)

# Display the dictionary with execution times
print("\nFunction execution times:")
display(function_execution_times)
```
