Create a dictionary comprehension that squares each number in a given list and uses the number as the key.  

```python
lis=[1,2,3,4,5,6,7,8,9]

d={x:x**2 for x in lis}
d
```

Given two lists, one containing names and the other containing ages, create a dictionary comprehension to map names to ages.

```python
names=['murali','ganesh','seeram']
age=[22,23,24]

d={x:y for x,y in zip(names,age)}
d
```

Generate a dictionary comprehension that filters out names longer than 5 characters from a list of names.

```python
names = ['murali', 'ganesh', 'seeram']

result = {name: len(name) for name in names if len(name) >= 5}

print(result)
```

Create a dictionary comprehension to count the frequency of each character in a given string.

Given a list of tuples representing words and their lengths, create a dictionary comprehension to construct a dictionary where words are keys and lengths are values.

```python
word_lengths = [('apple', 5), ('banana', 6), ('cat', 3)]

result = {word: length for word, length in word_lengths}

print(result)
```
