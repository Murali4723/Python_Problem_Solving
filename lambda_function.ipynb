How would you write a lambda function to sort a list of tuples based on the second
element in Python? Can you provide an example?


```python
# A list of tuples containing (Name, Age)
employees = [("Alice", 35), ("Bob", 23), ("Charlie", 28), ("Diana", 40)]

# Sort the list based on the second element (Index 1: Age)
sorted_employees = sorted(employees, key=lambda worker: worker[1])

print(sorted_employees)
# Output: [('Bob', 23), ('Charlie', 28), ('Alice', 35), ('Diana', 40)]
```

How can you create a lambda function to filter even numbers from a list in Python?
Can you give an example?

```python
lis = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Use lambda inside a filter
a = list(filter(lambda x: x % 2 == 0, lis))

print(a)
# Output: [2, 4, 6, 8]
```

Can you provide a lambda function with a conditional expression to categorize ages in
Python?

```python
# Define the lambda function with nested conditions
categorize_age = lambda age: "Child" if age < 13 else ("Teenager" if age < 20 else ("Adult" if age < 65 else "Senior"))

# 1. Test with individual values
print(categorize_age(8))   # Output: Child
print(categorize_age(15))  # Output: Teenager
print(categorize_age(34))  # Output: Adult
print(categorize_age(70))  # Output: Senior

print("-" * 20)

# 2. Test with a list of ages using map()
ages_list = [5, 17, 28, 62, 75]
categories = list(map(categorize_age, ages_list))

print(categories)
# Output: ['Child', 'Teenager', 'Adult', 'Adult', 'Senior']
```

How can you write a lambda function to calculate the factorial of a number using
recursion in Python? Can you demonstrate this with an example?

```python
fact = lambda n: 1 if n == 0 else n * fact(n - 1)

# Test the function
print(fact(5))  # Output: 120 (5 * 4 * 3 * 2 * 1)
print(fact(0))  # Output: 1
```
