Count the occurrences of an element in a tuple:

```python
my_tuple = (1, 2, 3, 4, 2, 5, 6, 2, 7)
for i in my_tuple:
  c=my_tuple.count(i)
  print(f'{i} {c}')

```

Reverse a TUple

```python
my_tuple = (1, 2, 3, 4, 2, 5, 6, 2, 7)

# To reverse a tuple, you can use slicing `[::-1]` which creates a new reversed tuple.
reversed_tuple = my_tuple[::-1]
print(reversed_tuple)
```

Find the maximum and minimum elements in a tuple:  

```python
my_tuple = (1, 2, 3, 4, 2, 5, 6, 2, 7)

print(max(my_tuple))
print(min(my_tuple))
```

Sort a tuple of numbers in ascending order:  

```python
my_tuple = (1, 2, 3, 4, 2, 5, 6, 2, 7)
a=list(my_tuple)
a.sort()
print(a)
```

Check if a tuple contains only unique elements:

```python
my_tuple = (1, 2, 3, 4, 2, 5, 6, 2, 7)
a=set(my_tuple)
if my_tuple ==a:
  print('the elements are unique')
else:
  print('the elements are not unique')
```

Multiply a tuple to repeat its elements:  

```python
my_tuple = (1, 2, 3)
repeated_tuple = my_tuple * 3
print(repeated_tuple)

my_another_tuple = ('a', 'b')
repeated_another_tuple = my_another_tuple * 2
print(repeated_another_tuple)
```
