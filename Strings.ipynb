Given a string, display it in a pyramid format.

```python
text = "hello"

for i in range(1, len(text) + 1):
    print(" " * (len(text) - i) + " ".join(text[:i]))
```

 Encode String Write a program that encodes a string by replacing each letter with the next letter (i.e., replace 'a' with 'b', 'b' with 'c', ..., 'z' with 'a').  


```python
a='abcd'
b=''
for char in a:
  c=ord(char)
  b+=chr(c+1)
print(b)
```

Find All Occurrences of Substring

```python
import re

text = "banana"
sub = "an"

# Find non-overlapping occurrences
for match in re.finditer(sub, text):
    print(f"Found '{match.group()}' at index: {match.start()}")
```
