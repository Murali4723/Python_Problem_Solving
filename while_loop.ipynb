Implement a program to reverse a given number using a while loop.

```python
def reverse_number(num):
    reversed_num = 0

    # Run the loop until the number becomes 0
    while num > 0:
        # Extract the last digit
        last_digit = num % 10

        # Append the digit to the reversed number
        reversed_num = (reversed_num * 10) + last_digit

        # Remove the last digit using floor division
        num = num // 10

    return reversed_num

# Example usage
original = 12345
result = reverse_number(original)
print(f"Original: {original} | Reversed: {result}")
```

Create a program that converts a binary number (entered as a string) to its decimal equivalent using a while loop.

```python
binary = input("Enter a binary number: ")

decimal = 0
power = 0

i = len(binary) - 1

while i >= 0:
    digit = int(binary[i])
    decimal = decimal + digit * (2 ** power)
    power += 1
    i -= 1

print("Decimal =", decimal)
```
