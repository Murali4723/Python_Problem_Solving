```python
sentence = "Python is an amazing programming language"

# Splits the sentence into words and takes the 0th index of each word
first_letters = [word[0] for word in sentence.split()]

print(first_letters)
```

```python
numbers = [1, 2, 2, 3, 4, 4, 1, 5]

# Track seen items to ensure each element is only added once
seen = set()
unique_numbers = [x for x in numbers if not (x in seen or seen.add(x))]

print(unique_numbers)
# Output: [1, 2, 3, 4, 5]
```

```python
sentence = "Learning Python is fun"

# Splits into words, reverses the list using slicing [::-1]
reversed_words = [word for word in sentence.split()[::-1]]

# Optional: Join them back into a sentence
reversed_sentence = " ".join(reversed_words)

print(reversed_words)
# Output: ['fun', 'is', 'Python', 'Learning']
```
