Bank Account using class :

```python
class BankAccount:
    """A class representing a personal bank account with safety checks."""

    def __init__(self, account_holder, initial_balance=0.0):
        """Initializes the account with a holder name and starting balance."""
        self.account_holder = account_holder
        # Use an underscore prefix to indicate the balance should not be modified directly
        self._balance = float(initial_balance)
        self.transaction_history = [f"Account opened with balance: ₹{initial_balance:.2f}"]

    def get_balance(self):
        """Safely retrieves the current account balance."""
        return self._balance

    def deposit(self, amount):
        """Adds a positive amount to the account balance."""
        if amount <= 0:
            print("❌ Error: Deposit amount must be greater than zero.")
            return False

        self._balance += amount
        self.transaction_history.append(f"Deposited: ₹{amount:.2f}")
        print(f"✅ Successfully deposited ₹{amount:.2f}. New balance: ₹{self._balance:.2f}")
        return True

    def withdraw(self, amount):
        """Subtracts a positive amount if sufficient funds exist."""
        if amount <= 0:
            print("❌ Error: Withdrawal amount must be greater than zero.")
            return False
        if amount > self._balance:
            print(f"❌ Error: Insufficient funds. Available balance: ₹{self._balance:.2f}")
            return False

        self._balance -= amount
        self.transaction_history.append(f"Withdrew: ₹{amount:.2f}")
        print(f"✅ Successfully withdrew ₹{amount:.2f}. New balance: ₹{self._balance:.2f}")
        return True

    def display_statement(self):
        """Prints the full mini-statement for the account."""
        print(f"\n--- Statement for {self.account_holder} ---")
        for transaction in self.transaction_history:
            print(f"  • {transaction}")
        print(f"Final Balance: ₹{self._balance:.2f}\n----------------------------------")


# --- How to use the BankAccount class ---

# 1. Open a new account
my_account = BankAccount(account_holder="Arjun Sharma", initial_balance=5000.0)

# 2. Perform transactions
my_account.deposit(1500.50)
my_account.withdraw(2000.0)

# 3. Test edge case errors (unauthorized withdrawal amount)
my_account.withdraw(6000.0)

# 4. View transaction history
my_account.display_statement()
```

student class

```python
class student:
  def __init__(self,name,student_id):
    self.name=name
    self.student_id=student_id
    self.student_marks={}

  def subject_marks(self,subject_name,marks):
    self.subject_name=subject_name
    self.marks=marks
    self.student_marks[subject_name]=marks

  def average(self):
    total_score=sum(self.student_marks.values())
    return total_score/len(self.student_marks)

  def get_final_grade(self):
        """Determines the final letter grade based on the average score."""
        average = self.average()
        """Prints a well-formatted academic report card for the student."""
        print(f"\n=====================================")
        print(f"         ACADEMIC REPORT CARD        ")
        print(f"=====================================")
        print(f"Student Name : {self.name}")
        print(f"Student ID   : {self.student_id}")
        print(f"-------------------------------------")
        if average >= 90: return "A+"
        elif average >= 80: return "A"
        elif average >= 70: return "B"
        elif average >= 60: return "C"
        elif average >= 50: return "D"
        else: return "Fail"

  def display_report_card(self):
      """Prints a well-formatted academic report card for the student."""
      print(f"\n=====================================")
      print(f"         ACADEMIC REPORT CARD        ")
      print(f"=====================================")
      print(f"Student Name : {self.name}")
      print(f"Student ID   : {self.student_id}")
      print(f"-------------------------------------")
      print(f"-------------------------------------")
      avg = self.average()
      print(f"Average Score: {avg:.2f}%")
      print(f"Final Grade  : {self.get_final_grade()}")
      print(f"=====================================\n")

student1=student(name="Rohan Patel", student_id="STU202601")
student1.subject_marks("Mathematics", 92)
student1.subject_marks("Science", 85)
student1.subject_marks("English", 78)
student1.subject_marks("Computer Science", 95)

student1.display_report_card()
```

Create a class Rectangle that calculates the area and perimeter of a rectangle.

```python
class rectangle:
  def __init__(self,length,width):
    self.length=length
    self.width=width

  def area(self):
    return self.length*self.width

  def perimeter(self):
    return 2*(self.length+self.width)

p1=rectangle(length=5,width=5)
print(p1.area())
print(p1.perimeter())
```

11. Create a class Circle that calculates the area and circumference of a circle


```python
import math

class Circle:
    """A class to represent a circle and compute its properties."""

    def __init__(self, radius):
        """Initializes the circle with a given radius."""
        self.radius = radius

    def circumference(self):
        """Calculates and returns the circumference (2 * pi * r)."""
        return 2 * math.pi * self.radius

    def area(self):
        """Calculates and returns the area (pi * r^2)."""
        return math.pi * (self.radius ** 2)


# --- Testing the Circle class ---

# Create a circle object with a radius of 5
c1 = Circle(radius=5)

# Print calculations rounded to 2 decimal places
print(f"Radius:        {c1.radius}")
print(f"Circumference: {c1.circumference():.2f}")
print(f"Area:          {c1.area():.2f}")

```

Create a class TemperatureConverter that converts temperatures between Celsius and
Fahrenheit. Implement methods for conversion in both directions.

```python
class TemperatureConverter:
    """A class utility to convert temperatures between Celsius and Fahrenheit."""

    @staticmethod
    def celsius_to_fahrenheit(celsius):
        """Converts Celsius to Fahrenheit: (C * 9/5) + 32"""
        return (celsius * 9 / 5) + 32

    @staticmethod
    def fahrenheit_to_celsius(fahrenheit):
        """Converts Fahrenheit to Celsius: (F - 32) * 5/9"""
        return (fahrenheit - 32) * 5 / 9


# --- How to use the TemperatureConverter class ---

# Note: Because these methods use the @staticmethod decorator,
# you don't need to create an instance (like converter = TemperatureConverter())
# to use them. You can call them directly using the class name!

# 1. Convert Celsius to Fahrenheit
c_temp = 37.0  # Normal body temperature
f_result = TemperatureConverter.celsius_to_fahrenheit(c_temp)
print(f"{c_temp}°C is equal to {f_result:.1f}°F")

# 2. Convert Fahrenheit to Celsius
f_temp = 98.6
c_result = TemperatureConverter.fahrenheit_to_celsius(f_temp)
print(f"{f_temp}°F is equal to {c_result:.1f}°C")

# 3. Test another point (Freezing point of water)
freezing_f = TemperatureConverter.celsius_to_fahrenheit(0)
print(f"0°C is equal to {freezing_f}°F")
```
