Given a string, swap the characters at positions i and j  Example: Input: s = "hello", i = 1, j = 3  Output: "hlleo"

```python
s = "hello"

i = 1
j = 3

s = list(s)

s[i], s[j] = s[j], s[i]

s = "".join(s)

print(s)
```

Given a list, swap elements at even indices with elements at odd indices.

```python
numbers = [10, 20, 30, 40, 50, 60]

for i in range(0, len(numbers) - 1, 2):
    numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

print(numbers)
```

Given two lists of equal length, swap the i-th element of the first list with the i-th element of the second list.

```python

list1 = [1, 2, 3, 4]
list2 = [10, 20, 30, 40]

for i in range(len(list1)):
    list1[i], list2[i] = list2[i], list1[i]

print(list1)
print(list2)
```

Given two lists, swap their middle elements. If a list has an even number of elements, consider the latter middle.  

```python
list1 = [1, 2, 3, 4, 5]
list2 = [10, 20, 30, 40, 50]

i = len(list1) // 2
j = len(list2) // 2

list1[i], list2[j] = list2[j], list1[i]

print(list1)
print(list2)
```

Given a list, repeatedly swap the largest number in the list with its right neighbor until it reaches the end of the list.  

```python
lst = [3, 8, 2, 5, 9, 1]

max_index = lst.index(max(lst))

while max_index < len(lst) - 1:
    lst[max_index], lst[max_index + 1] = lst[max_index + 1], lst[max_index]
    max_index += 1

print(lst)
```
