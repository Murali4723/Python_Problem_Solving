```python
#Write a Python program for a store that calculates discounts
#based on the purchase amount and membership status.
#Ask theuser for the purchase amount and
#whether they have a membership.
#If the purchase amount is greater than $100 and they havea membership, apply a 20% discount.
#If the purchase amount is greater than $100 without a membership, apply a 10% discount.
#Otherwise, no discount is applied.

user_purchase=int(input('enter your purchase amount: '))
membership_status=input('enter the membership status yes/no: ')

def discount_afterprice(n):
    discount=user_purchase*(n/100)
    price_afterdiscount=user_purchase-discount
    return price_afterdiscount

if user_purchase>100 and membership_status=='yes':
    a=discount_afterprice(20)
    print(a)

elif user_purchase>100 and membership_status=='no':
    b=discount_afterprice(10)
    print(b)
else:
    print(f'no discount applied final price is {user_purchase}')
```
