How can you create a function with a default parameter in Python? Can you provide
an example?

```python
def greet(name="Guest"):
    return f"Hello, {name}!"


# Calling without an argument uses the default value
print(greet())  # Output: Hello, Guest!

# Calling with an argument overrides the default value
print(greet("Alice"))  # Output: Hello, Alice!
```

How would you demonstrate enclosing scope in Python with an example

```python
def outer_function():
    message = "Hello from the enclosing scope!"  # Enclosing variable

    def inner_function():
        # The inner function can read the enclosing variable
        print(message)

    inner_function()


outer_function()
# Output: Hello from the enclosing scope!
```

Can you show how to access a global variable from within a function in Python?

```python
# Global variable
discount = 5


def apply_discount(price):
    # Directly reading 'discount' from global scope
    return price - discount


print(apply_discount(100))  # Output: 95
```

How can you modify a nonlocal variable in a nested function in Python? Can you give
an example?

```python
def outer_function():
    message = "Original Message"  # Nonlocal variable

    def inner_function():
        nonlocal message  # Bind 'message' to the enclosing scope
        message = "Modified Message!"  # Modify the variable

    inner_function()
    print("Outer function scope:", message)


outer_function()
# Output: Outer function scope: Modified Message!
```

How would you access a nonlocal variable in a nested function in Python? Can you
provide an example?

```python
def outer_function():
    # Variable in the enclosing (nonlocal) scope
    trainer_name = "Alex"

    def inner_function():
        # Directly accessing the nonlocal variable for reading
        print(f"The trainer assigned to you is {trainer_name}.")

    inner_function()


outer_function()
# Output: The trainer assigned to you is Alex.
```
