 Find the index of the last occurrence of an element in a list ?  

```python
my_list = ['a', 'b', 'c', 'a', 'b', 'c']
target = 'b'

# Search backwards using a generator
last_index = max(i for i, x in enumerate(my_list) if x == target)

print(last_index)
# Output: 4

```

Shuffle a list randomly:  

```python
import random

lst = [1, 2, 3, 4, 5]
random.shuffle(lst)  # Modifies in-place

print(lst)  # Output: [3, 1, 5, 2, 4] (randomized)
```

```python
from itertools import combinations

def get_all_subsets(lst):
    subsets = []
    for r in range(len(lst) + 1):
        for combo in combinations(lst, r):
            subsets.append(list(combo))
    return subsets

# Example
print(get_all_subsets([1, 2]))  # Output: [[], [1], [2], [1, 2]]
```
