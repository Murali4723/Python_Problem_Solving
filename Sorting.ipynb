Given a list of numbers, sort them in ascending order.  

```python
lis=[55,66,77,88,99,4,5,6,7,8,9,5,2,1,58,5852]
lis.sort()
print(lis)
```

Given a list of strings, sort them alphabetically.  

```python
lis=['g','a','b','e','q']
lis.sort()
print(lis)
```

Given a list of strings, sort them by their length.

```python
words = ["apple", "hi", "banana", "cat", "a"]

result = sorted(words, key=len)

print(result)
```

Given a list of numbers, sort them in descending order.

```python
lis=[55,66,77,88,99,4,5,6,7,8,9,5,2,1,58,5852]
lis.sort(reverse=True)
print(lis)
```

Given a list of tuples where each tuple contains two elements, sort the list based on the second element of each tuple.

```python
items = [(1, 5), (2, 3), (3, 8), (4, 1)]

result = sorted(items, key=lambda x: x[1])

print(result)

```

Given a list of words, sort them based on their last character.  

```python
words = ["apple", "hi", "banana", "cat", "a"]
result = sorted(words , key= lambda x:x[-1])
print(result)
```

Given a list of numbers, sort them based on their frequency in the list. If two numbers have the same frequency, sort them in ascending order.  

```python
from collections import Counter

nums = [4, 5, 6, 4, 5, 4, 2, 6, 2, 2]

freq = Counter(nums)

result = sorted(nums, key=lambda x: (-freq[x], x))

print(result)
```

Given a matrix (2D list), sort its rows based on the sum of elements in each row.  

```python
matrix=[[1,55,9],[5,66,8]]
result=sorted(matrix, key =sum)
print(result)
```

Given a list of numbers and a reference number, sort the list based on the absolute difference between each number and the reference number.  

```python
nums = [10, 5, 20, 15, 8]
ref = 12

result = sorted(nums, key=lambda x: abs(x - ref))

print(result)
```

 Given a list of integers and strings, sort them such that all strings are on the left and all integers are on the right.

```python
nums = [10, 5, 20, 15, 8, "apple", "hi", "banana", "cat", "a"]

result = sorted(nums, key=lambda x: (not isinstance(x, str), x))

print(result)
```
